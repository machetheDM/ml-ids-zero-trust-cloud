## 2. ML-IDS Pipeline Flowchart (Paper Figure 2)

```python
fig, ax = plt.subplots(figsize=(14, 4))
ax.set_xlim(0, 10); ax.set_ylim(0, 2); ax.axis('off')

stages = [
    ('Traffic\nCapture', '#3498db'),
    ('Feature\nExtraction', '#2ecc71'),
    ('SMOTE +\nNormalisation', '#f39c12'),
    ('RFECV\n(25 features)', '#9b59b6'),
    ('LSTM +\nRandom Forest', '#e74c3c'),
    ('ZTA Policy\nDecision', '#1abc9c'),
    ('Auto\nResponse', '#e67e22'),
    ('Active\nLearning', '#2c3e50'),
]

for i, (label, colour) in enumerate(stages):
    x = i * 1.25 + 0.2
    rect = mpatches.FancyBboxPatch((x, 0.4), 1.0, 1.0, boxstyle='round,pad=0.05',
                                    facecolor=colour, edgecolor='white', alpha=0.9)
    ax.add_patch(rect)
    ax.text(x + 0.5, 0.9, label, ha='center', va='center', fontsize=8.5,
            color='white', fontweight='bold', wrap=True)
    if i < len(stages) - 1:
        ax.annotate('', xy=(x + 1.25, 0.9), xytext=(x + 1.0, 0.9),
                    arrowprops=dict(arrowstyle='->', color='#7f8c8d', lw=2))

# Feedback loop arrow
ax.annotate('', xy=(0.7, 0.35), xytext=(9.45, 0.35),
            arrowprops=dict(arrowstyle='->', color='#95a5a6', lw=1.5,
                            connectionstyle='arc3,rad=-0.3'))
ax.text(5.1, 0.1, 'Continuous feedback loop (active learning)', ha='center', fontsize=8, color='#95a5a6')

ax.set_title(
    'End-to-End ML-IDS Pipeline (Paper Figure 2)\n'
    'From traffic capture to automated ZTA response orchestration\n'
    'Machethe (2026), ECCU500, EC-Council University',
    fontsize=10, pad=10,
)
plt.tight_layout()
plt.savefig('../results/figures/05_mlids_pipeline.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 3. Threat Taxonomy Heatmap (Paper Table 2)

```python
import pandas as pd

threats = ['Network Intrusion', 'Lateral Movement', 'Data Exfiltration', 'Insider Threat', 'Zero-Day Exploit', 'DDoS/DoS', 'Misconfiguration']
pillars = ['Identity', 'Devices', 'Networks', 'Applications', 'Data']

# Severity mapping: 3=CRITICAL, 2=HIGH, 1=MEDIUM, 0=not primary
severity = np.array([
    [0, 0, 3, 2, 0],   # Network Intrusion
    [3, 2, 2, 0, 0],   # Lateral Movement
    [0, 0, 1, 1, 3],   # Data Exfiltration
    [3, 1, 0, 0, 1],   # Insider Threat
    [0, 0, 1, 3, 1],   # Zero-Day Exploit
    [0, 0, 3, 2, 0],   # DDoS/DoS
    [1, 2, 1, 1, 1],   # Misconfiguration
])

fig, ax = plt.subplots(figsize=(9, 6))
cmap = sns.color_palette(['#ecf0f1', '#f39c12', '#e74c3c', '#8e44ad'], as_cmap=True)
sns.heatmap(severity, annot=True, fmt='d', cmap='RdYlGn_r',
            xticklabels=pillars, yticklabels=threats,
            linewidths=0.5, linecolor='white', ax=ax,
            cbar_kws={'label': '0=None | 1=MEDIUM | 2=HIGH | 3=CRITICAL'})
ax.set_title('Threat Category × ZTA Pillar Severity Matrix\n(Paper Table 2 — Machethe 2026, ECCU500)', fontsize=11)
ax.set_xlabel('ZTA Pillar (NIST SP 800-207)'); ax.set_ylabel('Threat Category')
plt.tight_layout()
plt.savefig('../results/figures/05_threat_taxonomy_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 4. Model Selection by ZTA Pillar (Paper §7 Recommendations)

```python
pillar_model_map = {
    'Identity':     ['LSTM', 'UEBA Clustering'],
    'Devices':      ['Random Forest', 'Behaviour Baseline'],
    'Networks':     ['LSTM + RF Ensemble', 'XGBoost', 'SVM'],
    'Applications': ['Autoencoder', 'XGBoost'],
    'Data':         ['Autoencoder', 'Isolation Forest'],
}

pillar_colours = {'Identity': '#3498db', 'Devices': '#2ecc71', 'Networks': '#e67e22', 'Applications': '#9b59b6', 'Data': '#e74c3c'}

fig, ax = plt.subplots(figsize=(12, 6))
ax.axis('off')

for i, (pillar, models) in enumerate(pillar_model_map.items()):
    x_base = i * 2.2 + 0.3
    # Pillar header box
    rect = mpatches.FancyBboxPatch((x_base, 4.0), 1.8, 0.8,
                                    boxstyle='round,pad=0.05',
                                    facecolor=pillar_colours[pillar], edgecolor='white')
    ax.add_patch(rect)
    ax.text(x_base + 0.9, 4.4, pillar, ha='center', va='center',
            fontsize=10, fontweight='bold', color='white')
    # Model boxes
    for j, model in enumerate(models):
        y = 3.0 - j * 1.1
        rect2 = mpatches.FancyBboxPatch((x_base, y), 1.8, 0.8,
                                         boxstyle='round,pad=0.05',
                                         facecolor=pillar_colours[pillar], edgecolor='white', alpha=0.3)
        ax.add_patch(rect2)
        ax.text(x_base + 0.9, y + 0.4, model, ha='center', va='center',
                fontsize=9, color='#2c3e50')
        ax.annotate('', xy=(x_base + 0.9, y + 0.8), xytext=(x_base + 0.9, 4.0),
                    arrowprops=dict(arrowstyle='->', color=pillar_colours[pillar], lw=1.2))

ax.set_xlim(0, 11.5); ax.set_ylim(0.5, 5.5)
ax.set_title(
    'ML Model Deployment by ZTA Pillar\n'
    'Paper §7 Recommendations | NIST SP 800-207 | Machethe (2026), ECCU500',
    fontsize=11, y=1.01,
)
plt.tight_layout()
plt.savefig('../results/figures/05_model_by_pillar.png', dpi=150, bbox_inches='tight')
plt.show()
```
