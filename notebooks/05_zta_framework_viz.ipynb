## Introduction: Zero Trust Architecture and ML-IDS Integration

**Zero Trust Architecture (ZTA)** is a security paradigm defined in **NIST Special Publication 800-207** (Rose et al., 2020) that eliminates implicit trust from all network communication. The foundational principle is *"never trust, always verify"*: every access request must be authenticated, authorised, and continuously validated regardless of its origin.

Traditional perimeter-based security models assume that threats originate externally and that internal network traffic can be trusted. ZTA rejects this assumption entirely. With cloud infrastructure, remote work, and lateral movement attacks commonplace, insider threats and compromised credentials can propagate freely inside a trusted perimeter. ZTA addresses this by enforcing least-privilege access per-session, continuously monitoring all east-west traffic, and automating policy enforcement.

### Five ZTA Pillars (NIST SP 800-207)

| Pillar | Core Function | ML Integration |
|--------|--------------|----------------|
| **Identity** | Continuous authentication, MFA, UEBA | UEBA clustering + LSTM temporal anomaly detection |
| **Devices** | Device trust scoring, endpoint compliance | Isolation Forest for baseline deviation |
| **Networks** | Micro-segmentation, east-west monitoring | Random Forest + XGBoost + LSTM ensemble |
| **Applications** | API security, zero-day detection | Autoencoder unsupervised anomaly detection |
| **Data** | Data classification, exfiltration prevention | Ensemble methods (Autoencoder + RF) |

### ML-IDS Performance (NSL-KDD, 10-fold CV)

| Model | Accuracy | FPR |
|-------|----------|-----|
| LSTM | 98.1% | 1.8% |
| XGBoost | 97.3% | 1.9% |
| Random Forest | 96.8% | 2.1% |
| SVM | 94.2% | 3.4% |
| Autoencoder | 91.5% | 4.2% |

```python
import sys, os
sys.path.insert(0, '..')

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.ticker as mticker
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

os.makedirs('../results/figures', exist_ok=True)

ZTA_BLUE = '#1F4E79'
ML_GREEN = '#117A65'
GREY     = '#7f8c8d'

# -----------------------------------------------------------------------
# Cell 2 — ZTA Pillar Diagram
# 5 pillars as styled boxes, ML method per pillar, arrows, 300 DPI
# Save: results/figures/07_zta_framework.png
# -----------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(15, 8))
ax.set_xlim(0, 15); ax.set_ylim(0, 9); ax.axis('off')

ax.text(7.5, 8.5, 'Zero Trust Architecture: ML-IDS Integration',
        ha='center', va='center', fontsize=15, fontweight='bold', color=ZTA_BLUE)
ax.text(7.5, 8.0, 'NIST SP 800-207 | Machethe (2026), ECCU500 | EC-Council University',
        ha='center', va='center', fontsize=10, color='#555555')

pillar_data = [
    ('Identity',     'UEBA + LSTM\nTemporal Anomaly',         '#1F4E79'),
    ('Devices',      'Isolation Forest\nBehaviour Baseline',  '#2E75B6'),
    ('Networks',     'Random Forest +\nXGBoost + LSTM',       '#117A65'),
    ('Applications', 'Autoencoder\nAnomaly Detection',        '#B7950B'),
    ('Data',         'Ensemble Methods\n(AE + RF)',            '#C0392B'),
]

x_starts = [0.4, 3.2, 6.0, 8.8, 11.6]
bw, bh_zta, bh_ml = 2.3, 1.5, 1.3

for (pillar, ml_method, colour), x in zip(pillar_data, x_starts):
    # ZTA pillar box (top)
    rect_zta = mpatches.FancyBboxPatch((x, 5.6), bw, bh_zta,
        boxstyle='round,pad=0.09', facecolor=ZTA_BLUE, edgecolor='white', lw=2)
    ax.add_patch(rect_zta)
    ax.text(x + bw/2, 6.55, pillar,
            ha='center', va='center', fontsize=11, fontweight='bold', color='white')
    ax.text(x + bw/2, 6.15, 'ZTA Pillar',
            ha='center', va='center', fontsize=8, color='#b3cde0')

    # Arrow ZTA -> ML
    ax.annotate('', xy=(x + bw/2, 4.3), xytext=(x + bw/2, 5.6),
                arrowprops=dict(arrowstyle='->', color=GREY, lw=2.0))

    # ML method box (bottom)
    rect_ml = mpatches.FancyBboxPatch((x, 2.9), bw, bh_ml,
        boxstyle='round,pad=0.09', facecolor=ML_GREEN, edgecolor='white', lw=2)
    ax.add_patch(rect_ml)
    for j, line in enumerate(ml_method.split('\n')):
        ax.text(x + bw/2, 3.75 - j * 0.42, line,
                ha='center', va='center', fontsize=8.5, color='white', fontweight='bold')

    # Dashed arrow ML -> ZTA enforcement (upward feedback)
    ax.annotate('', xy=(x + bw/2, 5.6), xytext=(x + bw/2, 4.22),
                arrowprops=dict(arrowstyle='->', color=ML_GREEN, lw=1.5, ls='dashed'))
    ax.text(x + 0.06, 5.0, 'enforces', fontsize=7, color=ML_GREEN, style='italic')

# ZTA enforcement banner
banner = mpatches.FancyBboxPatch((0.2, 1.8), 14.6, 0.85,
    boxstyle='round,pad=0.08', facecolor='#D6EAF8', edgecolor=ZTA_BLUE, lw=1.5)
ax.add_patch(banner)
ax.text(7.5, 2.22,
        'ZTA Policy Engine: "Never trust, always verify"  (NIST SP 800-207, Rose et al., 2020)',
        ha='center', va='center', fontsize=10, color=ZTA_BLUE, fontweight='bold')

# Legend
legend_handles = [
    mpatches.Patch(facecolor=ZTA_BLUE, label='ZTA Enforcement Pillar'),
    mpatches.Patch(facecolor=ML_GREEN, label='ML Detection Method'),
]
ax.legend(handles=legend_handles, loc='lower right', fontsize=9,
          framealpha=0.95, edgecolor='#cccccc')

plt.tight_layout()
path07 = '../results/figures/07_zta_framework.png'
fig.savefig(path07, dpi=300, bbox_inches='tight')
print(f'Saved [300 DPI] -> {path07}')
plt.show()
```

STAGE_COLOURS = ['#1F4E79', '#2E75B6', '#117A65', '#B7950B', '#C0392B']

for i, ((title, sub), colour) in enumerate(zip(stages, STAGE_COLOURS)):

        ax.annotate('', xy=(x + bw + gap, y_c), xytext=(x + bw, y_c),

```python
# -----------------------------------------------------------------------
# Cell 4 — Risk Score Distribution Simulation
# 1000 simulated network flows: normal (mean=25, std=10) + attack (mean=75, std=15)
# Threshold line at score=75
# Save: results/figures/08_risk_score_dist.png
# -----------------------------------------------------------------------
rng = np.random.RandomState(42)
n_samples = 1000

normal_scores = rng.normal(loc=25, scale=10, size=n_samples)
attack_scores = rng.normal(loc=75, scale=15, size=n_samples)

# Clip to [0, 100]
normal_scores = np.clip(normal_scores, 0, 100)
attack_scores = np.clip(attack_scores, 0, 100)

sns.set_theme(style='darkgrid', font_scale=1.05)
fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
fig.suptitle(
    'Network Flow Risk Score Distribution — Simulated (n=1,000 each class)\n'
    'Machethe (2026), ECCU500 | Threshold = 75 (95th-pct normal baseline)',
    fontsize=12, fontweight='bold',
)

THRESHOLD = 75

for ax, scores, label, colour, fill_c in zip(
    axes,
    [normal_scores, attack_scores],
    ['Normal Traffic', 'Attack Traffic'],
    [ML_GREEN, '#C0392B'],
    ['#abebc6', '#f5b7b1'],
):
    ax.hist(scores, bins=40, color=colour, edgecolor='white', alpha=0.82, label=label)
    ax.axvline(THRESHOLD, color='#E74C3C', lw=2.5, ls='--', label=f'Threshold = {THRESHOLD}')
    ax.set_xlabel('Risk Score [0 - 100]', fontsize=11)
    ax.set_ylabel('Frequency', fontsize=11)
    ax.set_title(f'{label}\nmean={scores.mean():.1f}, std={scores.std():.1f}',
                 fontsize=10, fontweight='bold')
    ax.legend(fontsize=9)
    ax.set_xlim(-5, 110)

# Combined overlay plot
fig2, ax2 = plt.subplots(figsize=(9, 5))
ax2.hist(normal_scores, bins=40, color=ML_GREEN, edgecolor='white', alpha=0.65,
         label='Normal (mean=25, std=10)')
ax2.hist(attack_scores, bins=40, color='#C0392B', edgecolor='white', alpha=0.65,
         label='Attack (mean=75, std=15)')
ax2.axvline(THRESHOLD, color='#E74C3C', lw=2.5, ls='--', label=f'Decision Threshold = {THRESHOLD}')

above = np.sum(attack_scores >= THRESHOLD)
below = np.sum(normal_scores >= THRESHOLD)
ax2.text(78, ax2.get_ylim()[1] * 0.7 if ax2.get_ylim()[1] > 0 else 60,
         f'Detected attacks:\n{above/n_samples*100:.1f}% above threshold',
         fontsize=9, color='#C0392B', fontweight='bold')
ax2.text(5, ax2.get_ylim()[1] * 0.7 if ax2.get_ylim()[1] > 0 else 60,
         f'False positives:\n{below/n_samples*100:.1f}% normal\nabove threshold',
         fontsize=9, color=ML_GREEN, fontweight='bold')

ax2.set_xlabel('Risk Score [0 - 100]', fontsize=11)
ax2.set_ylabel('Frequency', fontsize=11)
ax2.set_title(
    'Network Flow Risk Score Distribution (Overlaid)\n'
    'Machethe (2026), ECCU500 | Threshold = 75',
    fontsize=11, fontweight='bold',
)
ax2.legend(fontsize=9)
ax2.set_xlim(-5, 110)
plt.tight_layout()

path08 = '../results/figures/08_risk_score_dist.png'
fig2.savefig(path08, dpi=300, bbox_inches='tight')
print(f'Saved [300 DPI] -> {path08}')
plt.show()
```

# Cell 5 — ZTA Implementation Roadmap (Gantt Chart)
# 6 phases with durations from paper §5.5 / Table 4
# Save: results/figures/09_implementation_gantt.png
# -----------------------------------------------------------------------
import matplotlib.dates as mdates
from datetime import datetime, timedelta

phases = [
    ('Phase 1: Foundation',          1,  4,  '#1F4E79'),
    ('Phase 2: Data Collection',      5,  8,  '#2E75B6'),
    ('Phase 3: Model Development',    9,  14, '#117A65'),
    ('Phase 4: Integration',          15, 18, '#B7950B'),
    ('Phase 5: Validation',           19, 22, '#C0392B'),
    ('Phase 6: Operationalisation',   23, 30, '#6C3483'),
]

sns.set_theme(style='whitegrid', font_scale=1.0)
fig, ax = plt.subplots(figsize=(13, 6))

start_ref = datetime(2026, 1, 5)  # Monday week 1

for i, (name, w_start, w_end, colour) in enumerate(phases):
    d_start = start_ref + timedelta(weeks=w_start - 1)
    d_end   = start_ref + timedelta(weeks=w_end)
    duration = (d_end - d_start).days
    bar_start = mdates.date2num(d_start)

    ax.barh(i, duration, left=bar_start, height=0.55,
            color=colour, edgecolor='white', lw=1.5, alpha=0.88)
    ax.text(bar_start + duration / 2,
            i,
            f'  {name}  (Wks {w_start}-{w_end})',
            va='center', ha='center',
            fontsize=8.5, fontweight='bold', color='white')

ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
ax.xaxis.set_major_locator(mdates.MonthLocator())
fig.autofmt_xdate(rotation=30, ha='right')

ax.set_yticks(range(len(phases)))
ax.set_yticklabels([p[0] for p in phases], fontsize=9)
ax.invert_yaxis()
ax.set_xlabel('Timeline (2026)', fontsize=11)
ax.set_title(
    'ZTA ML-IDS Implementation Roadmap — Six-Phase Plan\n'
    'Paper §5.5, Table 4 | Machethe (2026), ECCU500 | EC-Council University',
    fontsize=12, fontweight='bold',
)
ax.grid(axis='x', alpha=0.4)

# Milestones
milestones = [
    (start_ref + timedelta(weeks=8),  'Data ready'),
    (start_ref + timedelta(weeks=14), 'Models trained'),
    (start_ref + timedelta(weeks=18), 'Deployed'),
    (start_ref + timedelta(weeks=22), 'Validated'),
]
for m_date, m_label in milestones:
    ax.axvline(mdates.date2num(m_date), color='#E74C3C', lw=1.2, ls='--', alpha=0.6)
    ax.text(mdates.date2num(m_date) + 0.5, len(phases) - 0.3,
            m_label, fontsize=7.5, color='#E74C3C', rotation=90, va='bottom')

plt.tight_layout()
path09 = '../results/figures/09_implementation_gantt.png'
fig.savefig(path09, dpi=300, bbox_inches='tight')
print(f'Saved [300 DPI] -> {path09}')
plt.show()
